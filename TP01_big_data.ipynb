```python
import requests
from bs4 import BeautifulSoup
import pandas as pd

# قائمة لتجميع كل البيانات
all_quotes_data = []

# الرابط الأساسي للموقع التجريبي
base_url = "http://quotes.toscrape.com"
next_page_url = "/"

# حلقة تكرارية للتنقل بين كل الصفحات حتى تنتهي
while next_page_url:
    # جلب الصفحة الحالية
    response = requests.get(base_url + next_page_url)
    soup = BeautifulSoup(response.text, 'html.parser')

    # استخراج كل الاقتباسات في الصفحة الحالية
    quotes_divs = soup.find_all('div', class_='quote')
    for quote in quotes_divs:
        text = quote.find('span', class_='text').text
        author = quote.find('small', class_='author').text
        tags = ", ".join([tag.text for tag in quote.find_all('a', class_='tag')])

        # إضافة البيانات في شكل قاموس (Dictionary)
        all_quotes_data.append({
            'Quote': text,
            'Author': author,
            'Tags': tags
        })

    # البحث عن زر الصفحة التالية (Next)
    next_button = soup.find('li', class_='next')
    if next_button:
        next_page_url = next_button.find('a')['href']
    else:
        next_page_url = None # التوقف عند الوصول لآخر صفحة

# 1. تحويل القائمة إلى Dataframe باستخدام Pandas
df = pd.DataFrame(all_quotes_data)

# 2. حفظ البيانات في ملف CSV
df.to_csv('quotes_demo.csv', index=False, encoding='utf-8-sig')

print(f"تم جمع إجمالي {len(df)} اقتباس ونقلها لـ CSV بنجاح!")
# عرض أول 5 أسطر من الجدول
df.head()
```

```python
import requests
from bs4 import BeautifulSoup

# 1. تحديد رابط الصفحة الأولى من موقع الكتب
url = "http://books.toscrape.com/catalogue/page-1.html"

# 2. إرسال طلب الاتصال
response = requests.get(url)

if response.status_code == 200:
    print("تم الاتصال بموقع الكتب بنجاح!\n")

    # 1. عرض كود الـ HTML الخام أولاً (أول 500 حرف)
    print("=== 1. كود الـ HTML الخام (Raw HTML) ===")
    print(response.text[:500])
    print("\n" + "="*50 + "\n")

    # 2. تحويل النص إلى Parser باستخدام BeautifulSoup
    soup = BeautifulSoup(response.text, 'html.parser')

    # 3. عرض كود الـ HTML المنسق والمحلل (Parsed HTML)
    print("=== 2. كود الـ HTML بعد التحليل والتنسيق (Parsed HTML) ===")
    print(soup.prettify()[:500])

else:
    print(f"فشل الاتصال، كود الحالة: {response.status_code}")
```

```python
# استخراج كتل الكتب من الصفحة الأولى
articles = soup.find_all('article', class_='product_pod')

print(f"عدد الكتب في هذه الصفحة: {len(articles)}\n")

for article in articles[:5]:  # عرض أول 5 كتب فقط للتجربة
    title = article.h3.a['title']
    price = article.find('p', class_='price_color').text
    rating = article.p['class'][1]
    availability = article.find('p', class_='instock availability').text.strip()

    print(f"عنوان الكتاب: {title}")
    print(f"السعر: {price}")
    print(f"التقييم: {rating}")
    print(f"التوفر: {availability}")
    print("-" * 50)
```

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd

books_data = []

# الحلقة التكرارية لجلب الـ 50 صفحة
for page_num in range(1, 51):
    url = f"http://books.toscrape.com/catalogue/page-{page_num}.html"
    response = requests.get(url)

    if response.status_code == 200:
        soup = BeautifulSoup(response.text, 'html.parser')
        articles = soup.find_all('article', class_='product_pod')

        for article in articles:
            title = article.h3.a['title']
            price = article.find('p', class_='price_color').text
            rating = article.p['class'][1]
            availability = article.find('p', class_='instock availability').text.strip()

            books_data.append({
                'Title': title,
                'Price': price,
                'Rating': rating,
                'Availability': availability
            })

        print(f"تم جلب الصفحة {page_num}/50 (الإجمالي حتى الآن: {len(books_data)} كتاب)")

# 1. تحويل القائمة إلى DataFrame
df = pd.DataFrame(books_data)

# 2. حفظ البيانات في ملف CSV
df.to_csv('books_1000.csv', index=False, encoding='utf-8-sig')

print("\n" + "="*50)
print(f"تم الإنجاز بنجاح! إجمالي عدد الأسطر في ملف CSV: {len(df)}")
print("="*50)

# عرض أول 5 أسطر من البيانات
df.head()
```

```python
import requests
from bs4 import BeautifulSoup

# 1. رابط الصفحة الأولى من منتجات Catalog108
url = "https://practice.scrapingcentral.com/products?page=1"

# 2. إرسال الطلب عبر requests
response = requests.get(url)

if response.status_code == 200:
    print("تم الاتصال بـ Catalog108 بنجاح!\n")

    # عرض الـ Raw HTML (كود خام)
    print("=== 1. كود الـ HTML الخام (Raw HTML) ===")
    print(response.text[:500])
    print("\n" + "="*50 + "\n")

    # التحويل إلى Parsed HTML باستعمال BeautifulSoup
    soup = BeautifulSoup(response.text, 'html.parser')

    # عرض الـ Parsed HTML (كود منظم ومحلل)
    print("=== 2. كود الـ HTML المُحلَّل (Parsed HTML) ===")
    print(soup.prettify()[:500])
else:
    print(f"فشل الاتصال، كود الحالة: {response.status_code}")
```

```python
# البحث عن عناصر المنتجات داخل الصفحة
products = soup.find_all('div', class_='product-card')

# إذا كان الـ Class مختلفاً في الموقع، يقرأ العامة من البطاقات
if not products:
    products = soup.find_all('div', class_='border')

print(f"عدد المنتجات المستخرجة في هذه الصفحة: {len(products)}\n")

for product in products[:5]:  # معاينة أول 5 منتجات
    # استخراج اسم المنتج
    name_elem = product.find(['h2', 'h3', 'a'])
    name = name_elem.text.strip() if name_elem else "N/A"

    # استخراج السعر
    price_elem = product.find(class_=lambda x: x and 'price' in x.lower())
    price = price_elem.text.strip() if price_elem else "N/A"

    print(f"اسم المنتج: {name}")
    print(f"السعر: {price}")
    print("-" * 40)
```

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd

catalog_data = []

# الدوران على جميع الصفحات حتى تنتهي المنتجات تماماً
page = 1
while True:
    url = f"https://practice.scrapingcentral.com/products?page={page}"
    response = requests.get(url)

    # التوقف إذا وصلنا لصفحة غير موجودة أو انتهت الصفحات
    if response.status_code != 200:
        print(f"انتهت الصفحات عند الصفحة {page - 1}")
        break

    soup = BeautifulSoup(response.text, 'html.parser')

    # البحث عن بطاقات المنتجات
    products = soup.find_all(['div', 'article'], class_=lambda x: x and ('product' in x.lower() or 'card' in x.lower()))

    # إذا لم نجد منتجات في الصفحة الحالية، نتوقف تلقائياً
    if not products:
        print(f"لا توجد منتجات إضافية. تم الوصول لنهاية الكتالوج عند الصفحة {page - 1}.")
        break

    for product in products:
        name_elem = product.find(['h2', 'h3', 'a'])
        price_elem = product.find(class_=lambda x: x and 'price' in x.lower())

        if name_elem:
            catalog_data.append({
                'Product_Name': name_elem.text.strip(),
                'Price': price_elem.text.strip() if price_elem else "N/A",
                'Page_Source': page
            })

    print(f"تم جلب الصفحة {page} (الإجمالي الكلي حتى الآن: {len(catalog_data)} منتج)")
    page += 1

# 1. إنشاء الـ DataFrame لكافة البيانات
df_all = pd.DataFrame(catalog_data)

# 2. حفظ كافة البيانات المجمعة في ملف CSV
df_all.to_csv('catalog108_full.csv', index=False, encoding='utf-8-sig')

print("\n" + "="*50)
print(f"تم الاكتفاء والاستخراج الكامل بنجاح!")
print(f"إجمالي عدد الأسطر/المنتجات المكتشفة والمحفوظة: {len(df_all)}")
print("="*50)

# عرض أول 5 أسطر لمعاينة النتيجة
df_all.head()
```
